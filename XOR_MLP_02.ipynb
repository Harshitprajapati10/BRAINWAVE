```python
import torch
import torch.nn as nn
import pytorch_lightning as L
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
```

```python
X = torch.tensor([
    [0., 0.],
    [0., 1.],
    [1., 0.],
    [1., 1.]
])

y = torch.tensor([[0.], [1.], [1.], [0.]])
```

```python
dataset = TensorDataset(X, y)
loader = DataLoader(dataset, batch_size=4)
```

```python
class XOR_MLP(L.LightningModule):
  def __init__(self):
    super().__init__()
    self.model = nn.Sequential(
        nn.Linear(2, 8),
          nn.ReLU(),
          nn.Linear(8, 4),
          nn.ReLU(),
          nn.Linear(4, 1),
          nn.Sigmoid()
    )
    self.loss_fn = nn.BCELoss()


  def forward(self, x):
    return self.model(x)

  def training_step(self, batch, batch_idx):
    x,y = batch
    y_hat = self(x)
    loss = self.loss_fn(y_hat, y)
    self.log("train_loss", loss)
    return loss

  def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=0.1)
```

```python
model = XOR_MLP()
trainer = L.Trainer(
    max_epochs=500,
    enable_checkpointing=True,
    logger=False
)
```

```python
trainer.fit(model, loader)
```

```python
trainer.save_checkpoint("xor_model.ckpt")
```

```python
loaded_model = XOR_MLP.load_from_checkpoint("xor_model.ckpt")
loaded_model.eval()
```

```python
def predict(inpt):
  with torch.no_grad():
    prediction = loaded_model(inpt)
    prediction = torch.round(prediction)
  print("Prediction:", int(prediction.item()))

predict(torch.tensor([[0., 0.]]))
predict(torch.tensor([[0., 1.]]))
predict(torch.tensor([[1., 0.]]))
predict(torch.tensor([[1., 1.]]))


```
